# CareerPilot AI — 5. Skill Gap Analysis

Compares a parsed resume against a job description and shows exactly which required skills the candidate already has and which ones they're missing.

**This step is fast and doesn't need Ollama** — it's a plain set comparison using the same skill keyword list from the other notebooks, so it works even while the LLM step in File 4 is slow or unavailable. An optional one-line LLM suggestion is offered at the end, but the core gap analysis never depends on it.

**Requires:** `1_resume_parsing.ipynb` has been run at least once (so `careerpilot.db` has a profile).

**Works with any job description** — paste one directly below, or auto-pull your top Adzuna match from `top_matches.json` if you've run File 3.


## Step 1 — Setup


```
import json
import sqlite3

DB_PATH = "careerpilot.db"
TOP_MATCHES_PATH = "top_matches.json"

# Same list as 1_resume_parsing.ipynb (Step 4) and 2_job_fetching.ipynb —
# keep all three in sync if you add a skill.
SKILLS = [
    "Python", "Java", "JavaScript", "SQL", "React", "FastAPI", "Flask",
    "Machine Learning", "Deep Learning", "AWS", "Docker", "Kubernetes",
    "Git", "spaCy", "MongoDB", "PostgreSQL", "TensorFlow", "PyTorch",
    "Node.js", "HTML", "CSS", "C++", "Data Analysis", "NLP",
]

def get_skills(text: str) -> list:
    text_lower = text.lower()
    return sorted({skill for skill in SKILLS if skill.lower() in text_lower})

print("Ready.")
```

## Step 2 — Load the candidate's parsed resume


```
def load_latest_profile() -> dict:
    conn = sqlite3.connect(DB_PATH)
    try:
        cur = conn.cursor()
        cur.execute("SELECT * FROM career_profiles ORDER BY id DESC LIMIT 1")
        row = cur.fetchone()
        cols = [d[0] for d in cur.description]
    finally:
        conn.close()
    if row is None:
        raise ValueError("No profiles found — run 1_resume_parsing.ipynb first.")
    profile = dict(zip(cols, row))
    profile["skills"] = json.loads(profile["skills"]) if profile["skills"] else []
    return profile

profile = load_latest_profile()
print(f"Candidate: {profile.get('name') or profile.get('filename')}")
print(f"Resume skills: {profile['skills']}")
```

## Step 3 — Provide the job description

**Option A (default below):** paste any job description text directly — this is the general-purpose path, works for any posting from anywhere.

**Option B:** auto-pull your #1 Adzuna match instead — uncomment the block in the cell below if you'd rather compare against that.


```
# --- Option A: paste a job description here ---
job_title = "Example Role"
job_description_text = """
Paste the full job description text here, replacing this placeholder.
Include the required skills / responsibilities section for best results.
"""

# --- Option B: pull the top Adzuna match instead (uncomment to use) ---
# with open(TOP_MATCHES_PATH, "r") as f:
#     match_data = json.load(f)
# top_job = match_data["matches"][0]
# job_title = f"{top_job['title']} — {top_job['company']}"
# job_description_text = top_job["description"]

print(f"Analyzing against: {job_title}")
```

## Step 4 — Compute the skill gap


```
resume_skills = set(profile["skills"])
job_skills = set(get_skills(job_description_text))

matched_skills = sorted(resume_skills & job_skills)
missing_skills = sorted(job_skills - resume_skills)
extra_skills = sorted(resume_skills - job_skills)   # candidate has these, job doesn't ask for them

match_pct = round(len(matched_skills) / len(job_skills) * 100) if job_skills else 0

print(f"Job requires {len(job_skills)} tracked skills: {sorted(job_skills)}\n")
print(f"✅ Matched ({len(matched_skills)}): {matched_skills}")
print(f"❌ Missing ({len(missing_skills)}): {missing_skills}")
print(f"➕ Extra (not required, but candidate has): {extra_skills}\n")
print(f"Skill match: {match_pct}%")
```

## Step 5 (optional) — One-line suggestion via Ollama

Skippable — the gap analysis above already works without this. Only run this cell if Ollama is responding quickly right now; if it's slow, just use the missing-skills list from Step 4 directly, it's already the answer.


```
import time
import requests

OLLAMA_URL = "http://localhost:11434/api/generate"
OLLAMA_MODEL = "qwen3.5:0.8b"   # small + fast on purpose for this optional step

if missing_skills:
    prompt = (
        f"A candidate is missing these skills for a '{job_title}' role: {', '.join(missing_skills)}. "
        "In one sentence, suggest which single skill to learn first and why."
    )
    start = time.time()
    response = requests.post(OLLAMA_URL, json={
        "model": OLLAMA_MODEL, "prompt": prompt, "stream": False,
    }, timeout=60)
    print(f"(generated in {time.time() - start:.1f}s)")
    if response.status_code == 200:
        print(response.json().get("response", "").strip())
    else:
        print("Ollama error:", response.text)
else:
    print("No missing skills — candidate already covers everything tracked for this role.")
```
